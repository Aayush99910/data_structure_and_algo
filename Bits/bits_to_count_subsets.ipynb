# Finding subsets using bits
When you are given a list of set and you want to make a subsets that has all the items in that set.
You can use bit to do so as well. 
Whole concept is that lets say we have 4 bit representation of 0000, 0001 and when we do bit masking and we get a result that will determine if the binary number contains that item or not on from that set.

From what I have understood, lets say we have a set [1, 2]  
and we want to create all the subsets of this set  
we will go from 0 all the way to 3

Mask 0th --> 0000  
Mask 1st --> 0001  
Mask 2nd --> 0010  
Mask 3rd --> 0011  

For Mask 0:  
We do the BITWISE AND with 1 and 2  
so,  
0000 & 0001 = 0000 (This means no subset here so [])  
0000 & 0010 = 0000 (Same here)  
So, we add [] as our first subset  

For Mask 1:  
We do the BITWISE AND with 1 and 2  
so,  
0001 & 0001 = 0001 (This means 1 can be added in our subset [1])  
0001 & 0010 = 0000 (No match so dont add anything)  
So, we add [1] as our second subset  

For Mask 2:  
We do the BITWISE AND with 1 and 2  
so,  
0010 & 0001 = 0000 (No match)  
0010 & 0010 = 0010 (This means it matches second item and 2 can be added in our subset [2])  
So, we add [2] as our third subset  

For Mask 3:  
We do the BITWISE AND with 1 and 2  
so,  
0011 & 0001 = 0001 (This means 1 can be added in our subset [1])  
0011 & 0010 = 0010 (This means 2 can be added in our subset [1, 2])  
So, we add [1, 2] as our fourth subset  

Final answer  
--> [[], [1], [2], [1, 2]]


```python
def generate_subsets(s):
    n = len(s)
    all_subsets = []
    # we need to mask from 0 to left shit 1 by n 
    for mask in range(1 << n):
        subset = []
        # and do it for n times each loop
        for i in range(n):
            if (mask & (1 << i)):
                subset.append(s[i])
        all_subsets.append(subset)

    return all_subsets
```

```python
print("Generating subsets for [1, 2]:")
print(generate_subsets([1, 2]))

print("\nGenerating subsets for [1, 2, 3]:")
print(generate_subsets([1, 2, 3]))

print("\nGenerating subsets for [1, 2, 3, 4]:")
print(generate_subsets([1, 2, 3, 4]))
```
