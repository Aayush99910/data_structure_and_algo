Write a Python function that takes a binary sensor signal and returns the flipped version using
the NOT operator.
Hint:
1. Convert the binary signal to an integer.
220
2. Use the NOT operator to flip the bits.
3. Mask the result to ensure it matches the length of the original binary input.

```python
def flip_sensor_state(sensor_data_binary):
    # we first need to convert the given string to int base 2
    binary_number = int(sensor_data_binary, 2)

    # now we need to flip the data by using the ~ 
    not_binary_number = ~binary_number

    '''
    Now after doing this we need to make sure that we just get the last length digit 
    Example: 
        let's say I have 0b00010101 and if we do NOT of this then 
        NOT of 0b00010101 -> 0b11101010 this is wrong our output should be "0b1010" or 0b00001010

        for that we need to make a mask that makes the last length digit stay the same
        # If we use Bitwise AND then we can make sure that happens because only 1 AND 1 is 1 
        since 
        0b11101010
      & 0b00011111 
        0b00001010 -> our answer 
    So we need to get 0b00011111 to get this mask we need to left shift 1 by the given length of our binary number in string
    here that is 0b00100000 and then we subtract that by 1
    0b00100000
   -0b00000001
    0b00011111

    and now we do AND with mask and get the result
    '''

    # creating mask 
    mask = (1 << len(sensor_data_binary)) - 1

    # after doing this I can get the answer 
    result = not_binary_number & mask 

    # in short we can also do return ~binary_number & (1 << len(sensor_data_binary) - 1)
    return result 
```

```python
sensor_data = "10101"
print(bin(flip_sensor_state(sensor_data))) # Output: "0b1010"
```
